#### Analogia da Sacola de Supermercado

As **listas** são como sacolas de supermercado! Você coloca vários itens dentro de uma única sacola:

- **Sacola** = Lista
- **Itens** = Elementos da lista
- **Ordem** = Os itens ficam na ordem que você colocou

```python
# Exemplos de listas
estudantes = ['Fulano', 'João', 'Maria', 'Roberta']
frutas = ["maçã", "banana", "laranja", "uva"]
paises = ["Brasil", "Estados Unidos", "Japão", "Alemanha", "Portugal"]
numeros = [1, 5, 0]
```

#### Características das Listas:

- **Mutáveis**: Você pode adicionar, remover e modificar elementos
- **Ordenadas**: Os elementos ficam na ordem que você colocou
- **Indexadas**: Cada elemento tem uma posição (começando do 0!)
- **Flexíveis**: Podem conter qualquer tipo de dados

> 💡 **Dica importante:** Em Python, as listas são definidas com colchetes `[]` e os elementos são separados por vírgulas. Simples assim!

#### Criando Listas

A sintaxe é bem simples! Use colchetes `[]` e separe os elementos com vírgulas:

```python
# Lista de nomes (strings)
nomes = ['João', 'Maria', 'Pedro', 'Ana']

# Lista de números (misturados!)
numeros = [-10, 0.50, 4]

# Lista mista (qualquer tipo!)
miscelanea = ['Python', 42, True, 3.14]
```

#### Listas Vazias - Começando do Zero

Às vezes você quer criar uma lista vazia e ir adicionando elementos conforme necessário:

```python
# Criando uma lista vazia
lista_vazia = []
print(f"Lista inicial: {lista_vazia}")

# Adicionando elementos com append()
lista_vazia.append(10)
lista_vazia.append(20)
lista_vazia.append(30)
print(f"Lista após adicionar: {lista_vazia}")
```

**Resultado:**
```
Lista inicial: []
Lista após adicionar: [10, 20, 30]
```

> 🔓 **As listas são MUTÁVEIS!** Isso significa que você pode:
> - ✅ Adicionar elementos
> - ✅ Remover elementos  
> - ✅ Modificar elementos
> - ✅ Reorganizar elementos
> 
> É como uma sacola de supermercado onde você pode mexer à vontade!

#### Operações com Listas

Agora vamos aprender as operações essenciais para trabalhar com listas! São como as ferramentas da sua caixa de ferramentas de programação.

##### Percorrendo Listas (Iteração)

Uma das operações mais importantes é **percorrer** uma lista, ou seja, "visitar" cada elemento:

```python
frutas = ["maçã", "banana", "laranja", "uva"]

print("🍎 Lista de frutas:")
for fruta in frutas:
    print(f"- {fruta}")
```

**Resultado:**
```
🍎 Lista de frutas:

- maçã
- banana
- laranja
- uva
```

**Como funciona:**

- O `for` percorre cada elemento da lista
- A variável `fruta` assume o valor de cada elemento
- A cada iteração, executamos o código dentro do loop

> 💡 **Dica:** Percorrer listas com `for` é uma das operações mais comuns em Python!

##### Adicionando Elementos

Para adicionar elementos a uma lista, usamos o método `append()`:

```python
numeros = [1, 2, 3, 4]
print(f"Lista original: {numeros}")

numeros.append(5)
print(f"Após append(5): {numeros}")

numeros.append(6)
print(f"Após append(6): {numeros}")
```

**Resultado:**
```
Lista original: [1, 2, 3, 4]
Após append(5): [1, 2, 3, 4, 5]
Após append(6): [1, 2, 3, 4, 5, 6]
```

**Como funciona:**

- `append()` adiciona o elemento **no final** da lista
- A lista cresce automaticamente
- **Importante:** Não faça `lista = lista.append(item)` - isso está errado!

> 🔍 **Detalhe técnico:** O `append()` modifica a lista original. Não precisamos fazer `lista = lista.append(item)` porque a lista já é atualizada automaticamente.

##### Removendo Elementos

Para remover elementos de uma lista, usamos o método `remove()`:

```python
paises = ["Brasil", "EUA", "Japão", "Alemanha"]
print(f"Lista original: {paises}")

paises.remove("Japão")
print(f"Após remover 'Japão': {paises}")
```

**Resultado:**
```
Lista original: ['Brasil', 'EUA', 'Japão', 'Alemanha']
Após remover 'Japão': ['Brasil', 'EUA', 'Alemanha']
```

**Como funciona:**

- `remove()` remove o **primeiro elemento** que encontrar com o valor especificado
- Se o elemento não existir, Python dará erro
- A lista diminui automaticamente

> ⚠️ **Cuidado:** Se você tentar remover um elemento que não existe, Python dará um erro!

##### Ordenando Listas

Para ordenar uma lista, usamos o método `sort()`:

```python
numeros = [3, 1, 4, 2, 5]
print(f"Lista original: {numeros}")

numeros.sort()
print(f"Ordenada crescente: {numeros}")

numeros.sort(reverse=True)
print(f"Ordenada decrescente: {numeros}")
```

**Resultado:**
```
Lista original: [3, 1, 4, 2, 5]
Ordenada crescente: [1, 2, 3, 4, 5]
Ordenada decrescente: [5, 4, 3, 2, 1]
```

**Como funciona:**

- `sort()` ordena a lista **in-place** (modifica a lista original)
- Por padrão, ordena em ordem crescente
- `reverse=True` inverte a ordem (decrescente)

> 💡 **Dica:** `sort()` também funciona com strings (ordem alfabética)!

##### Acessando Elementos Específicos

Às vezes você quer acessar apenas um elemento específico da lista, não todos:

```python
frutas = ["maçã", "banana", "laranja", "uva"]

print(f"Primeira fruta: {frutas[0]}")  # índice 0
print(f"Segunda fruta: {frutas[1]}")   # índice 1
print(f"Terceira fruta: {frutas[2]}")  # índice 2
print(f"Última fruta: {frutas[3]}")    # índice 3
```

**Resultado:**
```
Primeira fruta: maçã
Segunda fruta: banana
Terceira fruta: laranja
Última fruta: uva
```

**Como funciona:**

- Usamos colchetes `[]` com o índice do elemento
- **Lembre-se:** Em Python, começamos a contar do **0**!
- `frutas[0]` = primeiro elemento
- `frutas[1]` = segundo elemento
- E assim por diante...

> 🔢 **Índices em Python:**
> 
> - Índice 0 = Primeiro elemento
> - Índice 1 = Segundo elemento  
> - Índice 2 = Terceiro elemento
> - E assim por diante...

##### Modificando Elementos

Para modificar um elemento específico da lista, usamos o índice:

```python
lista = [1, 2, 3, 4, 5]
print(f"Lista original: {lista}")

lista[2] = 10  # Modifica o terceiro elemento
print(f"Após modificar índice 2: {lista}")

lista[0] = 100  # Modifica o primeiro elemento
print(f"Após modificar índice 0: {lista}")
```

**Resultado:**
```
Lista original: [1, 2, 3, 4, 5]
Após modificar índice 2: [1, 2, 10, 4, 5]
Após modificar índice 0: [100, 2, 10, 4, 5]
```

**Como funciona:**

- Usamos `lista[índice] = novo_valor`
- O elemento na posição especificada é substituído
- A lista é modificada **in-place**

> 💡 **Dica:** Você pode modificar qualquer elemento que tenha um índice válido!

### Tuplas - O Ingresso de Cinema

As **tuplas** são como listas, mas com uma diferença fundamental: elas são **IMUTÁVEIS**! 

#### Analogia do Ingresso de Cinema

Imagine que você comprou um ingresso de cinema. No bilhete está escrito:

- **Filme:** "Vingadores"
- **Horário:** "20:30"
- **Sala:** "5"

Essas informações **não podem ser alteradas** depois que o ingresso foi impresso. É exatamente assim com as tuplas!

#### Diferença Principal: Mutável vs Imutável

| **Listas** | **Tuplas** |
|------------|------------|
| ✅ **Mutáveis** - podem ser alteradas | 🔒 **Imutáveis** - não podem ser alteradas |
| Uso: `[]` (colchetes) | Uso: `()` (parênteses) |
| Podem crescer/diminuir | Tamanho fixo |
| Flexíveis | Seguras |

#### Exemplos de Tuplas

```python
# Tupla de pessoas
pessoas = ("João", "Maria", "José")

# Tupla mista (diferentes tipos)
elementos = ("banana", 5, True, 3.14)

# Tupla de coordenadas
coordenadas = (40.7128, -74.0060)
```

#### Quando Usar Tuplas?

- **Coordenadas** (latitude, longitude)
- **Datas** (dia, mês, ano)
- **Configurações** que não mudam
- **Dados que devem ser protegidos** contra modificação acidental

> 💡 **Dica:** Use tuplas quando você tem dados que **nunca devem ser alterados**!

#### Criando Tuplas

Para criar uma tupla, use parênteses `()` e separe os elementos com vírgulas:

```python
# Tupla de coordenadas
coordenadas = (40.7128, -74.0060)

# Tupla de dias da semana
dias_da_semana = ("segunda", "terça", "quarta", "quinta", "sexta", "sábado", "domingo")

# Tupla de pontuação
pontuacao_jogo = (1200, 3)

# Tupla de informações pessoais
pessoa = ("João", 25, "São Paulo")
```

#### Tuplas Vazias

Sim, é possível criar tuplas vazias, mas são raramente úteis:

```python
tupla_vazia = ()
```

> 💡 **Dica:** Tuplas vazias são como ingressos em branco - tecnicamente existem, mas não servem para muito!

#### Operações com Tuplas

As tuplas têm operações similares às listas, mas lembre-se: **elas são imutáveis**!

##### Percorrendo Tuplas

Você pode percorrer tuplas da mesma forma que percorre listas:

```python
animais = ("gato", "cachorro", "papagaio", "peixe")

print("🐾 Lista de animais:")
for animal in animais:
    print(f"- {animal}")
```

**Resultado:**
```
🐾 Lista de animais:

- gato
- cachorro
- papagaio
- peixe
```

##### Verificando se um Elemento Existe

Para verificar se um elemento está na tupla, use `in`:

```python
carros = ("fusca", "fiesta", "oggi", "uno")

if "fusca" in carros:
    print("🚗 O fusca está na lista!")
else:
    print("❌ O fusca não está na lista!")
```

**Resultado:**
```
🚗 O fusca está na lista!
```

**Como funciona:**

- `in` verifica se o elemento existe na tupla
- Retorna `True` se encontrar, `False` se não encontrar
- Funciona igual com listas!

##### Adicionando e Removendo Elementos

**⚠️ ATENÇÃO:** Tuplas são **IMUTÁVEIS**! Você não pode adicionar ou remover elementos diretamente.

Mas há uma solução: **criar uma nova tupla**!

#### Adicionando Elementos (Criando Nova Tupla)

```python
tupla1 = (1, 2, 3)
print(f"Tupla original: {tupla1}")

# Para adicionar, criamos uma nova tupla
tupla2 = tupla1 + (4,)  # Note a vírgula!
print(f"Nova tupla: {tupla2}")
```

**Resultado:**
```
Tupla original: (1, 2, 3)
Nova tupla: (1, 2, 3, 4)
```

> ⚠️ **Importante:** A vírgula após o `4` é essencial! Sem ela, Python não reconhece como tupla.

#### Removendo Elementos (Criando Nova Tupla)

```python
tupla1 = (10, 20, 30, 40, 50)
print(f"Tupla original: {tupla1}")

# Para remover, criamos uma nova tupla sem os elementos indesejados
tupla2 = (tupla1[0], tupla1[2], tupla1[4])  # Pega apenas índices 0, 2 e 4
print(f"Nova tupla: {tupla2}")
```

**Resultado:**
```
Tupla original: (10, 20, 30, 40, 50)
Nova tupla: (10, 30, 50)
```

> 💡 **Dica:** Em vez de modificar, você sempre cria uma nova tupla!

##### Ordenando Tuplas

**⚠️ LEMBRE-SE:** Tuplas são **IMUTÁVEIS**! Não podemos ordenar diretamente.

**Solução:** Criar uma nova tupla ordenada!

```python
numeros = (5, 1, 3, 2, 4)
print(f"Tupla original: {numeros}")

# Ordenando criando nova tupla
numeros_ordenados = tuple(sorted(numeros))
print(f"Tupla ordenada: {numeros_ordenados}")
```

**Resultado:**
```
Tupla original: (5, 1, 3, 2, 4)
Tupla ordenada: (1, 2, 3, 4, 5)
```

**Como funciona:**

1. `sorted()` ordena e retorna uma **lista**
2. `tuple()` converte a lista em **tupla**

#### Conversões: Lista ↔ Tupla

```python
# Convertendo tupla em lista
tupla = (1, 2, 3)
lista = list(tupla)
print(f"Lista: {lista}")

# Convertendo lista em tupla
lista = [1, 2, 3]
tupla = tuple(lista)
print(f"Tupla: {tupla}")
```

**Resultado:**
```
Lista: [1, 2, 3]
Tupla: (1, 2, 3)
```

> 💡 **Dica:** Use `list()` para converter tupla em lista e `tuple()` para converter lista em tupla!

##### Acessando Elementos das Tuplas

Acesso a elementos funciona igual às listas! Vamos ver algumas técnicas:

#### Acesso por Índice

```python
tupla = ('a', 'b', 'c', 'd', 'e')

# Acessando elementos específicos
print(f"Primeiro elemento: {tupla[0]}")  # 'a'
print(f"Terceiro elemento: {tupla[2]}")  # 'c'
print(f"Último elemento: {tupla[-1]}")   # 'e'
```

**Resultado:**
```
Primeiro elemento: a
Terceiro elemento: c
Último elemento: e
```

#### Fatiamento (Slicing)

Para pegar uma "fatia" da tupla:

```python
tupla = ('a', 'b', 'c', 'd', 'e')

# Pegando elementos do índice 1 ao 3
subtupla = tupla[1:4]
print(f"Fatia [1:4]: {subtupla}")

# Pegando os primeiros 3 elementos
primeiros = tupla[:3]
print(f"Primeiros 3: {primeiros}")

# Pegando os últimos 2 elementos
ultimos = tupla[-2:]
print(f"Últimos 2: {ultimos}")
```

**Resultado:**
```
Fatia [1:4]: ('b', 'c', 'd')
Primeiros 3: ('a', 'b', 'c')
Últimos 2: ('d', 'e')
```

#### Índices Negativos

```python
tupla = ('a', 'b', 'c', 'd', 'e')

print(f"Último elemento: {tupla[-1]}")    # 'e'
print(f"Penúltimo elemento: {tupla[-2]}") # 'd'
print(f"Antepenúltimo elemento: {tupla[-3]}") # 'c'
```

> 💡 **Dica:** Índices negativos contam de trás para frente! `-1` = último, `-2` = penúltimo, etc.

### Dicionários - O Dicionário de Palavras

Os **dicionários** são como um dicionário de palavras! Você procura por uma **chave** (palavra) e encontra o **valor** (significado).

#### Problema: Sistema de Condomínio

Imagine que você precisa criar um sistema para um condomínio. Você **poderia** fazer assim:

```python
apartamentos = [100, 101, 102, 103]
nomes = ['Fulano', 'Ana', 'Roberta', 'Maria', 'Eduardo']
pets = [True, False, True]
```

**🚨 PROBLEMA:** Temos 4 apartamentos, mas 5 nomes! O Eduardo é de qual apartamento? E se alguém adicionar um nome a mais? Como saber se a ordem está correta?

#### Solução: Dicionários!

Os dicionários resolvem esse problema organizando as informações por **chave** e **valor**:

```python
apartamentos = {
    100: 'Ana',
    101: 'Roberta', 
    102: 'Fulano',
    103: 'Eduardo'
}
```

#### Como Funcionam os Dicionários?

- **Chave**: O que você procura (ex: número do apartamento)
- **Valor**: O que você encontra (ex: nome do morador)
- **Único**: Cada chave aparece apenas uma vez
- **Flexível**: Valores podem ser qualquer tipo de dado

#### Analogia do Dicionário de Palavras

| **Dicionário de Palavras** | **Dicionário Python** |
|----------------------------|------------------------|
| **Palavra** (chave) | **Chave** |
| **Significado** (valor) | **Valor** |
| Procura "casa" → encontra "lugar onde moramos" | Procura `100` → encontra `'Ana'` |

> 💡 **Dica:** Dicionários são perfeitos para associar informações relacionadas!

#### Criando Dicionários

A sintaxe é bem simples! Use chaves `{}` e separe chave e valor com dois pontos `:`:

```python
# Dicionário de informações pessoais
pessoa = {
    "nome": "João",
    "idade": 30,
    "cidade": "São Paulo"
}

# Dicionário de notas
notas = {
    "Matemática": 8.5,
    "Português": 7.0,
    "História": 9.2
}

# Dicionário de configurações
config = {
    "tema": "escuro",
    "idioma": "português",
    "notificações": True
}
```

#### Estrutura: Chave : Valor

- **Chave**: Sempre à esquerda dos dois pontos
- **Valor**: Sempre à direita dos dois pontos
- **Separador**: Vírgula entre cada par chave:valor

> 💡 **Dica:** A ordem das chaves não importa! Você pode organizar como quiser.

#### Dicionários Vazios

Para criar um dicionário vazio:

```python
# Forma 1: Chaves vazias
meu_dicionario = {}

# Forma 2: Função dict()
meu_dicionario = dict()
```

Ambas criam um dicionário vazio que pode ser preenchido depois!

#### Operações Essenciais com Dicionários

##### Percorrendo Dicionários

```python
pessoa = {
    "nome": "João",
    "idade": 30,
    "cidade": "São Paulo"
}

# Percorrer chaves
for chave in pessoa:
    print(chave)

# Percorrer valores
for valor in pessoa.values():
    print(valor)

# Percorrer chaves e valores
for chave, valor in pessoa.items():
    print(f"{chave}: {valor}")
```

##### Adicionando Elementos

```python
pessoa = {"nome": "João", "idade": 30}
pessoa["cidade"] = "São Paulo"  # Adiciona nova chave
print(pessoa)
```

##### Acessando Elementos

```python
pessoa = {"nome": "João", "idade": 30}

# Forma 1: Colchetes (pode dar erro se chave não existir)
print(pessoa["nome"])

# Forma 2: get() (retorna None se chave não existir)
print(pessoa.get("cidade", "Não informado"))
```

##### Modificando Elementos

```python
pessoa = {"nome": "João", "idade": 25}
pessoa["idade"] = 30  # Modifica valor existente
print(pessoa)
```

##### Removendo Elementos

```python
pessoa = {"nome": "João", "idade": 30, "cidade": "SP"}

# Remover com del
del pessoa["cidade"]

# Remover com pop (retorna o valor)
idade = pessoa.pop("idade")
print(f"Idade removida: {idade}")
```

**24. Sistema de Notas com Múltiplos Alunos**

```python
alunos = {}

def adicionar_nota(aluno, nota):
    if aluno not in alunos:
        alunos[aluno] = []
    alunos[aluno].append(nota)
    print(f"Nota {nota} adicionada para {aluno}")

def calcular_media(aluno):
    if aluno in alunos and alunos[aluno]:
        media = sum(alunos[aluno]) / len(alunos[aluno])
        return media
    return None

def relatorio_alunos():
    print("\n=== RELATÓRIO DE ALUNOS ===")
    for aluno, notas in alunos.items():
        media = calcular_media(aluno)
        print(f"{aluno}: {notas} - Média: {media:.2f}")

# Exemplo de uso
adicionar_nota("João", 8.5)
adicionar_nota("João", 7.0)
adicionar_nota("Maria", 9.2)
relatorio_alunos()
```

**Explicação linha por linha:**

- **Linha 1:** `alunos = {}` - Dicionário para armazenar alunos e suas listas de notas
- **Linha 3:** `def adicionar_nota(aluno, nota):` - Função para adicionar nota a um aluno
- **Linhas 4-5:** Verifica se aluno existe, senão cria lista vazia para ele
- **Linha 6:** `alunos[aluno].append(nota)` - Adiciona nota à lista do aluno
- **Linha 9:** `def calcular_media(aluno):` - Função para calcular média de um aluno
- **Linha 13:** `def relatorio_alunos():` - Função para gerar relatório de todos os alunos

**25. Sistema de Estoque**

```python
estoque = {
    "produto1": {"nome": "Notebook", "preco": 2500.00, "quantidade": 5},
    "produto2": {"nome": "Mouse", "preco": 50.00, "quantidade": 20},
    "produto3": {"nome": "Teclado", "preco": 150.00, "quantidade": 15}
}

def vender_produto(codigo, quantidade):
    if codigo in estoque:
        if estoque[codigo]["quantidade"] >= quantidade:
            estoque[codigo]["quantidade"] -= quantidade
            valor_total = estoque[codigo]["preco"] * quantidade
            print(f"Venda realizada: {quantidade}x {estoque[codigo]['nome']} = R$ {valor_total:.2f}")
        else:
            print("Quantidade insuficiente em estoque!")
    else:
        print("Produto não encontrado!")

def relatorio_estoque():
    print("\n=== RELATÓRIO DE ESTOQUE ===")
    for codigo, produto in estoque.items():
        print(f"{produto['nome']}: {produto['quantidade']} unidades - R$ {produto['preco']:.2f}")

# Exemplo de uso
relatorio_estoque()
vender_produto("produto1", 2)
vender_produto("produto2", 5)
relatorio_estoque()
```

**Explicação linha por linha:**

- **Linhas 1-5:** Cria dicionário de estoque com produtos aninhados (dicionário dentro de dicionário)
- **Linha 7:** `def vender_produto(codigo, quantidade):` - Função para processar venda
- **Linha 8:** `if codigo in estoque:` - Verifica se produto existe no estoque
- **Linha 9:** `if estoque[codigo]["quantidade"] >= quantidade:` - Verifica se há estoque suficiente
- **Linha 10:** `estoque[codigo]["quantidade"] -= quantidade` - Reduz quantidade em estoque
- **Linha 16:** `def relatorio_estoque():` - Função para gerar relatório do estoque
- **Linhas 20-22:** Exemplos de uso do sistema de estoque

```python
pessoas = {'João': 25, 'Maria': 30, 'José': 40}

for nome, idade in pessoas.items():
    print(f'{nome} tem {idade} anos.')
```

Nesse exemplo, a variável `pessoas` é um dicionário que armazena o nome e idade de três pessoas.
